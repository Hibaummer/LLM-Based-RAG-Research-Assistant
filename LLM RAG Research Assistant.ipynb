## 2. Import the Required Libraries

```python
import pymupdf
import pandas as pd
import faiss
import numpy as np
from langchain_text_splitters import RecursiveCharacterTextSplitter
from sentence_transformers import SentenceTransformer
```

## 3. Load the Research Paper

```python
doc = pymupdf.open("Research Paper.pdf")

print("Total pages:", len(doc))
```

## 4. Extract Text from Every Page

```python
pages = []

for page_num, page in enumerate(doc):
    text = page.get_text()

    pages.append({
        "page": page_num + 1,
        "text": text
    })

print("Total pages extracted:", len(pages))
```

## 5. Inspect the Extracted Text

```python
print(pages[0]["text"][:2000])
```

## 6. Identify Empty Pages

```python
empty_pages = [
    p["page"] for p in pages
    if not p["text"].strip()
]

print("Empty pages:", empty_pages)
```

## 7. Text Chunking

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separators=["\n\n", "\n", " ", ""]
)
```

```python
#Split the Extracted Text
chunks = []

for page in pages:
    page_chunks = text_splitter.split_text(page["text"])

    for chunk in page_chunks:
        chunks.append({
            "page": page["page"],
            "text": chunk
        })

print("Total chunks:", len(chunks))
```

## 8. Inspect the Generated Chunks

```python
print(chunks[0]["text"])
print("\nPage number:", chunks[0]["page"])
print("\nChunk length:", len(chunks[0]["text"]))
```

## 12. Generate Text Embeddings

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
chunk_texts = [chunk["text"] for chunk in chunks]

embeddings = model.encode(
    chunk_texts,
    show_progress_bar=True,
    convert_to_numpy=True
)

print("Embeddings shape:", embeddings.shape)
```

```python
#Inspect the first embedding
print(embeddings[0])
print("Vector dimensions:", len(embeddings[0]))
```

```python
#Validate the embeddings
print("Number of chunks:", len(chunks))
print("Number of embeddings:", len(embeddings))
print("Embedding dimensions:", embeddings.shape[1])
print("All embeddings finite:", __import__("numpy").isfinite(embeddings).all())
```

## 13.Create a Vector Database using FAISS

```python
embeddings = np.asarray(embeddings, dtype="float32")

print("Embeddings shape:", embeddings.shape)
print("Data type:", embeddings.dtype)
```

```python
# Normalize embeddings
faiss.normalize_L2(embeddings)

# Create the vector index
dimension = embeddings.shape[1]

index = faiss.IndexFlatIP(dimension)

# Store embeddings in the index
index.add(embeddings)

print("Total vectors stored:", index.ntotal)
```

```python
#save the vector database
faiss.write_index(index, "research_paper.index")

print("FAISS index saved successfully!")
```

```python
#Test semantic search
query = "What is the main objective of this research paper?"

query_embedding = model.encode(
    [query],
    convert_to_numpy=True
).astype("float32")

faiss.normalize_L2(query_embedding)

scores, indices = index.search(query_embedding, k=3)

for rank, idx in enumerate(indices[0], start=1):
    if idx != -1:
        print(f"\nResult {rank}")
        print("Similarity score:", scores[0][rank - 1])
        print("Page:", chunks[idx]["page"])
        print("Text:", chunks[idx]["text"])
```

```python
#Retrieve relevant chunks from your research paper
def retrieve_chunks(question, k=3):
    query_embedding = model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    faiss.normalize_L2(query_embedding)

    scores, indices = index.search(query_embedding, k)

    results = []

    for score, idx in zip(scores[0], indices[0]):
        if idx != -1:
            results.append({
                "page": chunks[idx]["page"],
                "text": chunks[idx]["text"],
                "score": float(score)
            })

    return results
```

```python
#Connect the retrieved chunks to Gemini
def answer_question(question):
    results = retrieve_chunks(question, k=3)

    if not results:
        return "No relevant passages found."

    context = "\n\n".join(
        f"Page {item['page']}:\n{item['text']}"
        for item in results
    )

    prompt = f"""
You are a research paper assistant.

Answer the question using only the provided context.
If the answer is not available in the context, say so.
Do not invent facts.
Include page references where possible.

Context:
{context}

Question:
{question}

Answer:
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    return response.text
```

```python
question = "What is the main objective of this research paper?"

answer = answer_question(question)

print(answer)
```

## 20.Add source citations with similarity scores

```python
def ask_research_assistant(question):
    # Retrieve relevant passages
    results = retrieve_chunks(question, k=3)

    if not results:
        print("No relevant passages found.")
        return

    # Generate the answer
    answer = answer_question(question)

    print("QUESTION")
    print(question)

    print("\nANSWER")
    print(answer)

    print("\nRETRIEVED SOURCES")
    for i, item in enumerate(results, start=1):
        print(f"\nSource {i}")
        print(f"Page: {item['page']}")
        print(f"Similarity score: {item['score']:.3f}")
        print(item['text'][:300], "...")
```

```python
ask_research_assistant(
    "How does self-attention help the Transformer overcome the limitations of RNNs?"
)
```
